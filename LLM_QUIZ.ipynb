### 17c.4 Linear Forward 의 학습 의미

- Q, K, V 의 weight 행렬 `W_Q`, `W_K`, `W_V` 는 *학습 대상* 이다 (`requires_grad=True`).
- 학습 초기에는 random 값이지만 loss backward 가 누적되며 Q 는 "지금 위치가 알고 싶은 것", K 는 "각 위치가 무슨 정보를 가졌는지", V 는 "그 위치에서 가져올 실제 값" 을 표현하도록 분화한다.
- 한 입력 x 에서 *세 개의 다른 의미 공간으로 projection* 하기 때문에, Q·K 점곱은 단순 내적이 아닌 *학습된 유사도* 가 된다.

## 28. 정답표 — 신규 섹션 (§16 ~ §27, §17b, §17c)

### §16 Attention 심화

- 16.1 ① `W_Q, W_K, W_V` ② `D_out` ③ Q="알고 싶어 하는 것", K="누가 무슨 정보를", V="content (값)"
- 16.2 ① `K` , `T` ② `d_k` ③ `-inf` (혹은 매우 큰 음수) ④ `-1` ⑤ `V`, `D`
- 16.3 예시 답: "Q·K 의 점곱은 차원 `d_k` 에 비례한 분산을 가져, `d_k` 가 클수록 score 의 분산이 커진다. 그대로 softmax 를 통과시키면 분포가 너무 sharp 해져 gradient 가 작은 위치로 거의 흐르지 않게 된다. `sqrt(d_k)` 로 나누면 score 의 분산을 1 근방으로 맞춰 softmax 가 안정적으로 학습된다."
- 16.4 (O, O, X)
- 16.5 `D`, `H, d_h`, `H`, `T`, `d_h`, `D`
- 16.6 정답 = **2**
- 16.7 (O, X, X)
- 16.8 코드:
  ```python
  nn.Linear(d_in, d_out, bias=qkv_bias)         # W_query/W_key/W_value 모두 동일
  diagonal=1                                    # 상삼각 1 부터 (대각 제외 위쪽만 mask)
  queries.view(b, num_tokens, self.num_heads, self.head_dim).transpose(1, 2)
  keys.transpose(2, 3)                          # [B, H, T, d_h] -> [B, H, d_h, T]
  attn_scores.masked_fill_(mask_bool, -torch.inf)
  attn_weights = torch.softmax(attn_scores / keys.shape[-1]**0.5, dim=-1)
  (attn_weights @ values).transpose(1, 2)
  context_vec.contiguous().view(b, num_tokens, self.d_out)
  ```

### ★ §17b Chapter 2 핵심 (Self-supervised Dataset + Embedding)

- 17b.1 `max_length`, `1`, `1`, target 은 input 보다 `1` 칸 뒤
- 17b.2 예시 답: "텍스트 그 자체가 input 과 target (다음 토큰) 의 쌍을 자동으로 만들어 주므로 사람의 별도 라벨 없이 학습 신호가 만들어진다."
- 17b.3 첫 인자 `vocab_size`, 두 번째 인자 `output_dim` (= D). 통과 후 shape `[B, T, output_dim (=D)]`. 내부 weight shape `[vocab_size, output_dim]`.
- 17b.4 첫 인자 `context_length`, 두 번째 인자 `output_dim` (= D). 입력은 `torch.arange(max_length)` (또는 `T`). 출력 shape `[T, D]` (= `[max_length, D]`).
- 17b.5 합산 화살표 위 `broadcasting`. 합산 방식은 `덧셈 (element-wise)`.
- 17b.6 (O, X, O, O, X, O)
- 17b.7 코드:
  ```python
  input_chunk  = token_ids[i : i + max_length]
  target_chunk = token_ids[i + 1 : i + max_length + 1]
  return self.input_ids[idx], self.target_ids[idx]

  token_embedding_layer = torch.nn.Embedding(vocab_size, output_dim)
  pos_embedding_layer   = torch.nn.Embedding(context_length, output_dim)

  token_embeddings = token_embedding_layer(inputs)              # [B, T, output_dim]
  pos_embeddings   = pos_embedding_layer(torch.arange(context_length))  # [T, D]
  input_embeddings = token_embeddings + pos_embeddings          # [B, T, D]
  ```
